# Python Data Structures: Lists, Tuples, Dictionaries, and Sets

## Introduction to Data Collections

In Python, you need to store collections of data, such as:
- Numbers (integers, floats)
- Strings (text)
- Booleans (True/False values)
- Mixed data types

Python provides several built-in data structures to organize and manage these collections efficiently. In this lesson, we'll explore:

1. **Lists** - Ordered, mutable collections
2. **Tuples** - Ordered, immutable collections  
3. **Dictionaries** - Key-value pairs for fast lookups
4. **Sets** - Unordered collections of unique elements

Let's start with Lists!

## 1. Python Lists - Dynamic Arrays

**What is a List?**
- A list is an ordered collection of items (elements)
- Lists are **mutable** (can be changed after creation)
- Elements can be of different data types
- Lists are defined using square brackets `[]`
- Elements are separated by commas

**Creating an Empty List:**

````python
# Creating an empty list using square brackets
my_list = []

# Let's check what we created
print("Empty list:", my_list)
print("Type of my_list:", type(my_list))
print("Length of my_list:", len(my_list))
````

### Adding Items to a List

**The `append()` Method:**
- `append()` adds a single item to the **end** of the list
- The list grows dynamically as you add elements
- Syntax: `list_name.append(item)`

**Example: Building a list step by step**

````python
# Adding elements one by one using append()
my_list.append(1)       # Add the number 1
print("After adding 1:", my_list)

my_list.append(2)       # Add the number 2  
print("After adding 2:", my_list)

my_list.append(3)       # Add the number 3
print("After adding 3:", my_list)

# Final result
print("Final list:", my_list)
````

### Real-World Example: Grocery Store Inventory

**Scenario:** Rahul works in a grocery store and needs to maintain a list of items.

Let's see how he can:
1. Create an initial inventory list
2. Add new items as they arrive
3. Combine different orders

````python
# Rahul's initial grocery inventory (created with items already in it)
items = ["Milk", "Eggs", "Bread"]
print("Initial inventory:", items)

# New items arrive - add them to the inventory
items.append("Butter")
print("After adding Butter:", items)

items.append("Cheese")
print("After adding Cheese:", items)

print("=" * 50)  # Separator line for clarity

# BONUS: List Concatenation (joining two lists)
# Rahul has two separate lists of numbers he wants to combine
numbers = [10, 20, 30]              # First list
new_numbers = [40, 50, 60]          # Second list

# Using the + operator to concatenate (join) lists
combined_numbers = numbers + new_numbers
print("Original numbers:", numbers)
print("New numbers:", new_numbers)  
print("Combined list:", combined_numbers)
````

### List Concatenation - Combining Lists

**Important Concept:** Lists are **concatenatable**, meaning you can join them together.

**Scenario Continues:** Rahul receives a new order with additional items. Instead of adding them one by one, he can combine the entire new list with his existing inventory.

**Methods to combine lists:**
1. **Using `+` operator** - Creates a new list
2. **Using `extend()` method** - Adds all items to existing list

````python
# Rahul receives a new order with different items
new_order = ["Biscuits", "Chips", "Soda"]
print("New order items:", new_order)
print("Current inventory:", items)

# Combine the existing inventory with the new order
total_inventory = items + new_order
print("Total inventory after combining:", total_inventory)

# Note: The original 'items' list remains unchanged
print("Original items list:", items)
print("Length of total inventory:", len(total_inventory))
````

## Accessing and Modifying List Items

### List Indexing
**Key Concepts:**
- Lists use **zero-based indexing** (first item is at index 0)
- You can access individual items using square brackets `[index]`
- You can modify items by assigning new values to specific indices

**Index positions:**
```
List: ["Milk", "Eggs", "Bread", "Butter", "Cheese", "Biscuits", "Chips", "Soda"]
Index:   0      1       2        3        4         5          6       7
```

````python
# Accessing the first item (index 0)
first_item = total_inventory[0]
print("First item in inventory:", first_item)

# Let's see the full inventory for reference
print("Full inventory:", total_inventory)
print("Item at index 0:", total_inventory[0])
print("Item at index 1:", total_inventory[1])
print("Item at index 2:", total_inventory[2])
````

````python
# Modifying list items (lists are mutable!)
print("Before modification:", total_inventory[0])

# Change the first item from "Milk" to "Curd"
total_inventory[0] = "Curd"

print("After modification:", total_inventory[0])
print("Updated inventory:", total_inventory)

# This demonstrates that lists are MUTABLE (changeable)
````

````python
# Working with 2D Lists (Lists containing other lists)
# This is like a table with rows and columns

# Each inner list represents: [Name, Age]
student_list = [["John", 20], ["Jane", 22], ["Doe", 21]]

print("2D Student List:", student_list)
print("Type:", type(student_list))

# Challenge: Change John's age from 20 to 24
# We need to access: student_list[0][1]
# [0] = first student (John's data)
# [1] = second element in John's data (his age)
````

````python
# Accessing John's data (the first student)
johns_data = student_list[0]
print("John's data:", johns_data)

# Breaking it down:
print("John's name:", student_list[0][0])  # First student, first element (name)
print("John's age:", student_list[0][1])   # First student, second element (age)
````

````python
# Modifying John's age in the 2D list
print("Before change:", student_list)

# Change John's age from 20 to 24
student_list[0][1] = 24

print("After change:", student_list)
print("John's new age:", student_list[0][1])
````

````python
# List and String Slicing - Getting portions of data
# Slicing syntax: [start:end] (end is not included)

name = "Farhan"
print("Original name:", name)

# Get characters from index 1 to 3 (not including 3)
name_slice = name[1:3]
print("Characters 1-2:", name_slice)  # Output: "ar"

print("=" * 40)

# List slicing works the same way
numbers_list = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print("Original list:", numbers_list)

# Get elements from index 2 to 5 (not including 5)
list_slice = numbers_list[2:5]
print("Elements 2-4:", list_slice)  # Output: [3, 4, 5]

# Other slicing examples:
print("First 3 elements:", numbers_list[:3])      # [1, 2, 3]
print("Last 3 elements:", numbers_list[-3:])      # [7, 8, 9]
print("Every 2nd element:", numbers_list[::2])    # [1, 3, 5, 7, 9]
````

````python
# String manipulation using slicing
name = "Farhan"
print("Original name:", name)

# Create a new name by changing the first letter
# "M" + everything from index 1 onwards
new_name = "M" + name[1:]
print("New name:", new_name)  # Output: "Marhan"

# Note: Strings are immutable, so we create a new string
print("Original name unchanged:", name)

# Understanding the slicing:
print("name[1:]:", name[1:])  # Everything from index 1 to end: "arhan"
````

````python
# List Re-assignment vs Modification
# Understanding the difference between changing list contents vs replacing the entire list

# Create a list with numbers
list_dummy = [1, 2, 3, 4, 5]
print("Original list:", list_dummy)

# COMPLETE REPLACEMENT: Assign a completely new list to the variable
list_dummy = ["a", "b"]  # This creates a new list object
print("After complete replacement:", list_dummy)

# This is different from modifying individual elements:
# list_dummy[0] = "a"  # This would modify the existing list
# list_dummy[1] = "b"  # This would modify the existing list
````

````python
# Creating lists with different data types
# Lists can contain strings, numbers, booleans, and even other lists!

subjects = ["physics", "calculus", "poetry", "history"]
print("School subjects:", subjects)
print("Number of subjects:", len(subjects))
print("First subject:", subjects[0])
print("Last subject:", subjects[-1])  # Negative indexing: -1 means last item
````

````python
# Create a list of numerical grades
# This demonstrates lists with integer data
grades = [98, 97, 85, 88]
print("Student grades:", grades)

# Useful operations with numerical lists:
print("Highest grade:", max(grades))
print("Lowest grade:", min(grades))
print("Average grade:", sum(grades) / len(grades))
print("Total points:", sum(grades))
````

## Common List Methods in Python
Here are some useful methods you can use with lists, along with example code for each.

````python
# Example list
fruits = ["apple", "banana", "cherry"]

# append(): Add an item to the end of the list
fruits.append("orange")
print("After append:", fruits)

# insert(): Insert an item at a given position
fruits.insert(1, "grape")
print("After insert:", fruits)

# remove(): Remove the first occurrence of a value
fruits.remove("banana")
print("After remove:", fruits)

# pop(): Remove and return item at given index (default last)
last_fruit = fruits.pop()
print("After pop:", fruits)
print("Popped item:", last_fruit)

# sort(): Sort the list in place
fruits.sort()
print("After sort:", fruits)

# reverse(): Reverse the elements of the list in place
fruits.reverse()
print("After reverse:", fruits)

# count(): Count occurrences of a value
count_apple = fruits.count("apple")
print("Count of 'apple':", count_apple)

# index(): Return first index of a value
index_cherry = fruits.index("cherry")
print("Index of 'cherry':", index_cherry)
````

````python
# Creating a tuple using parentheses
my_tuple = ('Milk', 'Eggs', 'Bread', 'Butter', 'Cheese', 'Biscuits', 'Chips', 'Soda')

print("My tuple:", my_tuple)
print("Type:", type(my_tuple))
print("Length:", len(my_tuple))

# Key difference from lists: Tuples use () instead of []
# Compare with a list:
my_list_version = ['Milk', 'Eggs', 'Bread', 'Butter', 'Cheese', 'Biscuits', 'Chips', 'Soda']
print("List version:", my_list_version)
print("List type:", type(my_list_version))
````

## 2. Python Tuples - Immutable Sequences

**What is a Tuple?**
- A tuple is an ordered collection of items (like a list)
- Tuples are **immutable** (cannot be changed after creation)
- Defined using parentheses `()` instead of square brackets
- Elements are separated by commas
- Perfect for storing related data that shouldn't change

**When to use Tuples:**
- Coordinates (x, y)
- RGB color values (red, green, blue)
- Database records
- Configuration settings
- Any data that should remain constant

**Creating a Tuple:**

````python
# Accessing tuple elements (same as lists - zero-based indexing)
print("First item:", my_tuple[0])
print("Second item:", my_tuple[1])
print("Last item:", my_tuple[-1])

# Slicing also works with tuples
print("First 3 items:", my_tuple[:3])
print("Items 2-4:", my_tuple[2:5])

# Let's try to modify a tuple (this will cause an error!)
# Uncomment the next line to see the error:
# my_tuple[0] = "Yogurt"  # This will raise a TypeError!
````

## Tuple Methods Cheat Sheet

Tuples are immutable sequences in Python. Here are the main methods and operations you can use with tuples:

- **count(x)**: Returns the number of times `x` appears in the tuple.
- **index(x)**: Returns the first index of value `x` in the tuple.

Other useful tuple operations:
- Access by index: `my_tuple[0]`
- Slicing: `my_tuple[1:4]`
- Concatenation: `my_tuple + another_tuple`
- Repetition: `my_tuple * 2`
- Membership test: `'Eggs' in my_tuple`
- Length: `len(my_tuple)`

> Note: Tuples do not support methods that modify them (like `append`, `remove`, or `sort`) because they are immutable.


````python
# Example usage of tuple methods and operations
my_tuple = ('Milk', 'Eggs', 'Bread', 'Butter', 'Cheese', 'Biscuits', 'Chips', 'Soda')

# count(): Count occurrences of a value
print(my_tuple.count('Eggs'))  # Output: 1

# index(): Find the index of a value
print(my_tuple.index('Bread'))  # Output: 2

# Access by index
print(my_tuple[0])  # Output: 'Milk'

# Slicing
print(my_tuple[1:4])  # Output: ('Eggs', 'Bread', 'Butter')

# Concatenation
another_tuple = ('Juice', 'Water')
print(my_tuple + another_tuple)

# Repetition
print(my_tuple * 2)

# Membership test
print('Eggs' in my_tuple)  # Output: True

# Length
print(len(my_tuple))  # Output: 8
````

## 3. Python Dictionaries - Key-Value Storage

**What is a Dictionary?**
- A collection of key-value pairs
- Keys are unique identifiers used to access values
- Dictionaries are **mutable** (can be modified)
- Defined using curly braces `{}`
- Format: `{key1: value1, key2: value2}`

**Why use Dictionaries?**
- Fast lookups by key name (instead of index position)
- Self-documenting code (keys describe the data)
- Perfect for representing real-world objects with properties

**Dictionary vs List:**
- **List:** Access by index position `my_list[0]`
- **Dictionary:** Access by key name `my_dict["name"]`

A dictionary in Python is a collection of key-value pairs. Each key is unique and is used to access its corresponding value. Dictionaries are mutable, meaning you can add, remove, or change items after creation. They are defined using curly braces `{}` with keys and values separated by a colon `:`.

Example:

```python
grocery_prices = {
    "Milk": 2.5,
    "Eggs": 3.0,
    "Bread": 2.0
}
print(grocery_prices["Eggs"])  # Output: 3.0
```

- Keys can be strings, numbers, or other immutable types.
- Values can be any data type.
- Dictionaries are very useful for fast lookups and organizing data by named keys.

````python
# Creating a dictionary to represent a student
# Format: {key: value, key: value}
student = {
    "Name": "Farhan",
    "Age": 43,
    "Class": "Python"
}

print("Student dictionary:", student)
print("Type:", type(student))

# Notice how the dictionary describes itself:
# - "Name" key tells us this value is a name
# - "Age" key tells us this value is an age
# - "Class" key tells us this value is a class name
````

````python
# Accessing dictionary values using keys
# Syntax: dictionary_name[key]

student_name = student["Name"]
print("Student's name:", student_name)

# You can also access other values:
print("Student's age:", student["Age"])
print("Student's class:", student["Class"])

# Compare this to list access:
# In a list, you'd use: my_list[0], my_list[1], my_list[2]
# In a dictionary, you use meaningful key names!
````

````python
# Dictionary methods for exploring content

# Get all the keys
all_keys = student.keys()
print("All keys:", list(all_keys))

# Get all the values  
all_values = student.values()
print("All values:", list(all_values))

# Get all key-value pairs
all_items = student.items()
print("All items:", list(all_items))

# Safe way to get a value (won't crash if key doesn't exist)
name = student.get("Name")
print("Name using get():", name)

# If key doesn't exist, get() returns None instead of crashing
missing = student.get("Grade")  # This key doesn't exist
print("Missing key result:", missing)
````

### Looping Through a Dictionary
You can loop through a dictionary in Python to access its keys, values, or key-value pairs.

````python
# Looping through dictionaries - 3 different ways

print("=== Method 1: Loop through keys ===")
for key in student:
    print(f"Key: {key}")

print("\n=== Method 2: Loop through values ===")
for value in student.values():
    print(f"Value: {value}")

print("\n=== Method 3: Loop through key-value pairs ===")
for key, value in student.items():
    print(f"{key}: {value}")

# Method 3 is most useful as you get both the key and value!
````

````python
# Deleting items from dictionaries - 2 methods

print("=== Method 1: Using 'del' statement ===")
student1 = {"Name": "Farhan", "Age": 43, "Class": "Python"}
print("Before deletion:", student1)

del student1["Age"]  # Remove the Age key-value pair
print("After deleting Age:", student1)

print("\n=== Method 2: Using 'pop()' method ===")
student2 = {"Name": "Farhan", "Age": 43, "Class": "Python"}
print("Before pop:", student2)

# pop() removes AND returns the value
removed_age = student2.pop("Age")
print("After pop:", student2)
print("Removed value was:", removed_age)

# pop() is useful when you need to know what was removed!
````

````python
# Working with mixed data types in dictionaries
# Dictionaries can store different types of values!

students = {
    "Akash": 1,           # String key, Integer value
    "Raffan": 2,          # String key, Integer value  
    "Sneha": "Teacher",   # String key, String value
    "Krishiv": 5          # String key, Integer value
}

print("Students dictionary:", students)

# Convert all values to a list
values_list = list(students.values())
print("All values as list:", values_list)
print("Type of values_list:", type(values_list))

# Get all keys as a list
keys_list = list(students.keys())
print("All keys as list:", keys_list)
````

````python
# Converting two lists into a dictionary using zip()
# This is a powerful technique for creating dictionaries!

# Two separate lists with related data
names = ["Akash", "Raffan", "Sneha", "Krishiv"]
values = [1, 2, "Teacher", 5]

print("Names list:", names)
print("Values list:", values)

# zip() pairs up elements from both lists
# dict() converts the pairs into a dictionary
combined_dictionary = dict(zip(names, values))

print("Combined dictionary:", combined_dictionary)

# How it works:
# zip(names, values) creates: [("Akash", 1), ("Raffan", 2), ("Sneha", "Teacher"), ("Krishiv", 5)]
# dict() converts these tuples into key-value pairs
````

````python
# ## 4. Python Sets - Collections of Unique Elements

# **What is a Set?**
# - An unordered collection of unique elements (no duplicates allowed)
# - Sets are mutable (can add/remove items)
# - Defined using curly braces {} or set() constructor
# - Perfect for removing duplicates and mathematical set operations

print("=== Creating Sets ===")
# Method 1: Using curly braces
fruits_set = {"apple", "banana", "cherry"}
print("Fruits set:", fruits_set)
print("Type:", type(fruits_set))

print("\n=== Adding Elements ===")
# Add a single element
fruits_set.add("orange")
print("After adding 'orange':", fruits_set)

# Try adding a duplicate (won't change the set)
fruits_set.add("apple")  # "apple" already exists
print("After adding duplicate 'apple':", fruits_set)

print("\n=== Removing Elements ===")
# Remove an element (will cause error if not found)
fruits_set.remove("banana")
print("After removing 'banana':", fruits_set)

# Alternative: discard() won't cause error if element not found
fruits_set.discard("grape")  # "grape" doesn't exist, but no error
print("After discarding 'grape' (not found):", fruits_set)
````

````python
# Set Operations - Mathematical Set Theory in Action!

print("=== Set Operations ===")
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print("Set A:", set_a)
print("Set B:", set_b)

print("\n--- Union (combines both sets, no duplicates) ---")
union_result = set_a | set_b  # Alternative: set_a.union(set_b)
print("A ∪ B (Union):", union_result)

print("\n--- Intersection (common elements only) ---")
intersection_result = set_a & set_b  # Alternative: set_a.intersection(set_b)
print("A ∩ B (Intersection):", intersection_result)

print("\n--- Difference (elements in A but not in B) ---")
difference_result = set_a - set_b  # Alternative: set_a.difference(set_b)
print("A - B (Difference):", difference_result)

print("\n--- Symmetric Difference (elements in either A or B, but not both) ---")
symmetric_diff = set_a ^ set_b  # Alternative: set_a.symmetric_difference(set_b)
print("A ⊕ B (Symmetric Difference):", symmetric_diff)
````

````python
# Practical Example: Removing Duplicates with Sets

print("=== Removing Duplicates ===")
# List with duplicate numbers
numbers_with_duplicates = [1, 2, 2, 3, 4, 4, 4, 5, 1]
print("Original list:", numbers_with_duplicates)

# Convert to set (removes duplicates) then back to list
unique_numbers = list(set(numbers_with_duplicates))
print("After removing duplicates:", unique_numbers)

print("\n=== Real-world Example: Email List Cleanup ===")
email_list = ["john@email.com", "jane@email.com", "john@email.com", "doe@email.com", "jane@email.com"]
print("Original email list:", email_list)
print("Number of emails:", len(email_list))

# Remove duplicate emails
unique_emails = list(set(email_list))
print("Unique emails:", unique_emails)
print("Number of unique emails:", len(unique_emails))
````

````python
# Final Exercise: Using All Data Structures Together!

print("=== Comprehensive Example: Student Management System ===")

# 1. List of students (ordered, mutable)
students_list = ["Alice", "Bob", "Charlie", "Diana"]
print("Students list:", students_list)

# 2. Tuple for a student's grades (ordered, immutable)
alice_grades = (95, 87, 92, 88)
print("Alice's grades (tuple):", alice_grades)

# 3. Dictionary for student information (key-value pairs)
student_info = {
    "Alice": {"age": 20, "major": "Computer Science"},
    "Bob": {"age": 19, "major": "Mathematics"},
    "Charlie": {"age": 21, "major": "Physics"},
    "Diana": {"age": 20, "major": "Computer Science"}
}
print("Student information (dictionary):", student_info)

# 4. Set for unique majors (no duplicates)
majors = {"Computer Science", "Mathematics", "Physics", "Computer Science"}
print("Unique majors (set):", majors)

print("\n=== Summary Statistics ===")
print(f"Total students: {len(students_list)}")
print(f"Alice's average grade: {sum(alice_grades) / len(alice_grades):.1f}")
print(f"Number of unique majors: {len(majors)}")
print(f"Computer Science students: {[name for name, info in student_info.items() if info['major'] == 'Computer Science']}")
````

## 📚 Summary: Python Data Structures Comparison

| Data Structure | Syntax | Ordered | Mutable | Duplicates | Use Case |
|----------------|--------|---------|---------|------------|----------|
| **List** | `[1, 2, 3]` | ✅ Yes | ✅ Yes | ✅ Yes | General purpose, dynamic arrays |
| **Tuple** | `(1, 2, 3)` | ✅ Yes | ❌ No | ✅ Yes | Immutable data, coordinates |
| **Dictionary** | `{"key": "value"}` | ✅ Yes* | ✅ Yes | ❌ No (keys) | Key-value lookups, objects |
| **Set** | `{1, 2, 3}` | ❌ No | ✅ Yes | ❌ No | Unique items, set operations |

*Dictionaries maintain insertion order (Python 3.7+)

### 🎯 Key Takeaways for Learners:

1. **Choose the right tool:** Each data structure has specific strengths
2. **Lists** - Your go-to for most collection needs
3. **Tuples** - When data shouldn't change (coordinates, config)
4. **Dictionaries** - When you need fast lookups by name/key
5. **Sets** - When uniqueness matters or you need set math
