```python
# Solutions — run to verify assertions

# 1. remove duplicates while preserving order
nums = [1, 2, 2, 3, 1, 4]
seen = set()
unique_ordered = [x for x in nums if not (x in seen or seen.add(x))]
print('unique_ordered:', unique_ordered)
assert unique_ordered == [1,2,3,4]

# 2. unique names case-insensitive
names = ['Alice', 'bob', 'ALICE', 'Bob', 'Charlie']
unique_count = len({n.lower() for n in names})
print('unique_count:', unique_count)
assert unique_count == 3

# 3. sort dict by descending score
scores = {'Alice': 88, 'Bob': 95, 'Charlie': 70}
sorted_desc = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)
print('sorted_desc:', sorted_desc)
assert sorted_desc[0][0] == 'Bob'

# 4. top-3 word frequencies
from collections import Counter
sentence = 'the quick brown fox jumps over the lazy dog the fox'
words = sentence.split()
top3 = Counter(words).most_common(3)
print('top3 words:', top3)
assert top3[0][0] == 'the'

print('\nAll solution assertions passed. Nice work!')
```

```python
# Strings — sequences of characters (immutable)

s = "hello, world"
print('lower:', s.lower())
print('split:', s.split(','))
print('replace:', s.replace('world', 'Python'))
print('substring [0:5]:', s[0:5])

# Strings behave like sequences: iterate, index, slice
for ch in s:
    pass  # you can iterate over characters

# Convert between types
joined = '-'.join(['2025','10','07'])
print('joined date:', joined)

# Common use: text processing and formatting.
```

```python
# Dictionaries — key/value mappings, unordered (insertion-ordered since Python 3.7)

student = {'name': 'Aisha', 'age': 20, 'courses': ['math', 'physics']}
print('student:', student)

# Access and mutate
print('name:', student['name'])
student['age'] = 21
student['grade'] = 'A'
print('after updates:', student)

# Iterate keys/values/items
for k, v in student.items():
    print(k, '->', v)

# Useful methods: get (with default), pop, keys, values
print('pop grade:', student.pop('grade'))
print('get missing with default:', student.get('gpa', 0.0))

# Use dicts for lookup tables and structured records.
```

```python
# Sets — unordered, mutable, no duplicates

s = {1, 2, 3}
print('set s:', s)

s.add(4)
print('after add:', s)

s.add(2)  # adding existing value does nothing
print('after add duplicate 2:', s)

# Useful operations: union, intersection, difference
a = {1, 2, 3}
b = {2, 3, 4}
print('union:', a | b)
print('intersection:', a & b)
print('difference a - b:', a - b)

# Use sets when uniqueness and fast membership checks are important.
```

```python
# Tuples — ordered, immutable, allow duplicates

point = (10, 20)
print('point:', point)

# Single-element tuple needs a trailing comma
single = (42,)
print('single-element tuple:', single)

# Tuples are immutable; you can't item-assign
try:
    point[0] = 5
except TypeError as e:
    print('immutable:', e)

# But can unpack
x, y = point
print('unpacked x,y:', x, y)

# Use tuples for fixed collections or mixed types (like records). 
```

```python
# Lists — ordered, mutable, allow duplicates

# Create lists
fruits = ['apple', 'banana', 'cherry']
print('fruits:', fruits)

# Indexing and slicing
print('first item:', fruits[0])
print('slice [1:]:', fruits[1:])

# Mutating
fruits.append('date')
fruits.insert(1, 'blueberry')
print('after append/insert:', fruits)

# Remove
fruits.remove('banana')  # removes by value
popped = fruits.pop()    # removes last and returns it
print('after remove/pop:', fruits, 'popped:', popped)

# Other useful methods
nums = [3, 1, 4, 1, 5]
nums.sort()
print('sorted nums:', nums)
print('list comprehension (squares):', [x*x for x in nums])

# Lists are great when you need an ordered, changeable collection.
```
